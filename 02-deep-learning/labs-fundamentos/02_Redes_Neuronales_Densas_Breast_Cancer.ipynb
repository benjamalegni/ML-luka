👉 El valor de loss (Train_Loss o Test_Loss)  no se interpreta como un porcentaje de error. En realidad, es una medida en escala logarítmica que evalúa qué tan buenas son las probabilidades que predice el modelo respecto a la etiqueta verdadera.

Por ejemplo:
- Si el modelo predice y=1 con probabilidad 0.9 y realmente era 1 → loss ≈ 0.105 (muy bajo).
- Si predice y=1 con probabilidad 0.51 y era 1 → loss ≈ 0.673 (más alto, aunque acertó).
- Si predice y=1 con probabilidad 0.9 pero era 0 → loss ≈ 2.302 (muy alto, porque se equivocó con mucha confianza).


💡 **Resumiendo:**
- El loss refleja qué tan confiado y acertado fue el modelo en sus predicciones.
- El accuracy solo dice si adivinó bien o mal (0 o 1), sin importar la probabilidad.

# Bonus Track: Entendiendo el Overfitting

### Dataset Sintético para Clasificación Binaria

En estas prácticas vamos a trabajar con un **dataset sintético** generado con la función 
`make_classification` de *Scikit-Learn*.  

- **Cantidad de ejemplos:** 200  
- **Cantidad de variables:** 20  
  - 15 informativas (aportan señal para diferenciar clases)  
  - 5 redundantes (aportan ruido al problema)  
- **Tarea:** Clasificación binaria (clases 0 y 1)  

El objetivo es predecir a qué clase pertenece cada ejemplo a partir de sus características.  

Al tratarse de un dataset pequeño y con cierta complejidad, resulta ideal para **experimentar con redes neuronales densas**, mostrando fenómenos como el **overfitting** y luego aplicar técnicas como **Dropout** para mejorar la capacidad de generalización del modelo.

---

## Práctica 6: Red neuronal para clasificación con dataset sintético

Vamos a desarrollar una red neuronal para realizar clasificación binaria sobre este dataset. Vamos a seguir utilizando ReLU en las capas ocultas y en la capa de salida vamos a usar la función lineal.

El modelo va a entrenarse con Adam como optimizador y evaluaremos su desempeño mediante la métrica accuracy, en entrenamiento y en testeo, para observar como generaliza.


```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, roc_auc_score
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

# 1. Dataset pequeño sintético
X, y = make_classification(n_samples=200, n_features=20, n_informative=15, 
                           n_redundant=5, n_classes=2, random_state=42)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Escalado
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# 2. Modelo
model_6a = Sequential([
    Dense(256, activation='relu', input_shape=(X_train.shape[1],)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

model_6a.compile(optimizer=Adam(1e-5), loss='binary_crossentropy', metrics=['accuracy'])

# 3. Entrenamiento con muchos epochs
model_6a.fit(X_train, y_train, epochs=100, batch_size=8, verbose=0)
```

```python
# 4. Evaluación
y_pred_train = (model_6a.predict(X_train) > 0.5).astype("int32")
y_pred_test  = (model_6a.predict(X_test) > 0.5).astype("int32")

acc_train = accuracy_score(y_train, y_pred_train)
acc_test  = accuracy_score(y_test, y_pred_test)

print("=== Resultados (Overfitting esperado) ===")
print(f"Train acc={acc_train:.4f}")
print(f"Test  acc={acc_test:.4f}")
```

- Entrenamiento: 80-85% acc → la red logra aprender patrones del set de entrenamiento.
- Test: 70-75% acc → caída importante, o sea, el modelo no generaliza bien.

***
### Probando la red usando Dropout 
Vamos a modificar la red neuronal previamente desarrollada para abordar el problema de Overfitting. Usaremos Dropout en las capas ocultas para simplificar el modelo.

```python
# Estas lineas de codigo realizan una limpieza del backend
import tensorflow as tf, gc
tf.keras.backend.clear_session()
gc.collect()
```

```python
from tensorflow.keras.layers import Dropout

# 2. Modelo (fuerzo a que se haga en CPU y no GPU)
with tf.device('/cpu:0'):
    model_6b = Sequential([
        Dense(256, activation='relu', input_shape=(X_train.shape[1],)),
        Dropout(0.5),
        Dense(128, activation='relu'),
        Dropout(0.2),
        Dense(64, activation='relu'),
        Dropout(0.1),
        Dense(32, activation='relu'),
        Dense(1, activation='sigmoid')
    ])

    model_6b.compile(optimizer=Adam(1e-5), loss='binary_crossentropy', metrics=['accuracy'])

    # 3. Entrenamiento con muchos epochs
    model_6b.fit(X_train, y_train, epochs=100, batch_size=8, verbose=0)
```

```python
# Evaluación
y_pred_train_d = (model_6b.predict(X_train) > 0.5).astype("int32")
y_pred_test_d  = (model_6b.predict(X_test) > 0.5).astype("int32")

acc_train_d = accuracy_score(y_train, y_pred_train_d)
acc_test_d  = accuracy_score(y_test, y_pred_test_d)

print("\n=== Resultados CON Dropout ===")
print(f"Train acc={acc_train_d:.4f}")
print(f"Test  acc={acc_test_d:.4f}")
```

En este ejercicio, la idea es que probemos con diferentes configuraciones de regularización para entender si logramos mejorar el Overfitting que se prenseta en la primera red neuronal desarrollada.
